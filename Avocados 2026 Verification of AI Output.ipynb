```python
# Import Packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
# Read both years
data25 = pd.read_csv("2025Data.csv")
data26 = pd.read_csv("2026Data.csv")
```

```python
# Keep Total U.S., Periods 1-32
d25 = data25[(data25["Geography"] == "Total U.S.") & (data25["Period"] <= 32)]
d26 = data26[(data26["Geography"] == "Total U.S.") & (data26["Period"] <= 32)]
```

```python
# Function for percent change
def pct_change(old, new):
    return (new / old - 1) * 100
```

```python
# Total volume
vol25 = d25["Total Bulk and Bags"].sum()
vol26 = d26["Total Bulk and Bags"].sum()
print("2025 volume:", vol25)
print("2026 volume:", vol26)
print("Volume change %:", pct_change(vol25, vol26))
```

```python
# Volume-weighted ASP
asp25 = (d25["ASP Current Year"] * d25["Total Bulk and Bags"]).sum() / vol25
asp26 = (d26["ASP Current Year"] * d26["Total Bulk and Bags"]).sum() / vol26
print("2025 weighted ASP:", asp25)
print("2026 weighted ASP:", asp26)
print("ASP change %:", pct_change(asp25, asp26))
```

```python
# Conventional and organic volume
for avocado_type in ["Conventional", "Organic"]:
    x25 = d25[d25["Type"] == avocado_type]["Total Bulk and Bags"].sum()
    x26 = d26[d26["Type"] == avocado_type]["Total Bulk and Bags"].sum()
    print(avocado_type, "volume change %:", pct_change(x25, x26))
```

```python
# Conventional and organic ASP
for avocado_type in ["Conventional", "Organic"]:
    x25 = d25[d25["Type"] == avocado_type]
    x26 = d26[d26["Type"] == avocado_type]
    asp25_type = x25["ASP Current Year"].mean()
    asp26_type = x26["ASP Current Year"].mean()
    print(avocado_type, "ASP change %:", pct_change(asp25_type, asp26_type))
```

```python
# Conventional share of total volume
conv25 = d25[d25["Type"] == "Conventional"]["Total Bulk and Bags"].sum()
conv26 = d26[d26["Type"] == "Conventional"]["Total Bulk and Bags"].sum()
print("2025 conventional share %:", conv25 / vol25 * 100)
print("2026 conventional share %:", conv26 / vol26 * 100)
```

```python
# Product growth and share
products = ["4046", "4225", "4770", "TotalBagged"]

for product in products:
    p25 = d25[product].sum()
    p26 = d26[product].sum()
    print(product, "total growth %:", pct_change(p25, p26))
    print(product, "2025 share %:", p25 / vol25 * 100)
    print(product, "2026 share %:", p26 / vol26 * 100)
```

```python
# Conventional product growth
c25 = d25[d25["Type"] == "Conventional"]
c26 = d26[d26["Type"] == "Conventional"]
```

```python
for product in products:
    p25 = c25[product].sum()
    p26 = c26[product].sum()
    print("Conventional", product, "growth %:", pct_change(p25, p26))
```

```python
# Regional conventional growth
regions = [
    "Southeast", "Midsouth", "California", "South Central",
    "Northeast", "Great Lakes", "Plains", "West"
]

for region in regions:
    r25 = data25[
        (data25["Geography"] == region) &
        (data25["Period"] <= 32) &
        (data25["Type"] == "Conventional")
    ]["Total Bulk and Bags"].sum()

    r26 = data26[
        (data26["Geography"] == region) &
        (data26["Period"] <= 32) &
        (data26["Type"] == "Conventional")
    ]["Total Bulk and Bags"].sum()

    print(region, "growth %:", pct_change(r25, r26))
```

```python
# Regional organic growth
regions = [
    "Southeast", "Midsouth", "California", "South Central",
    "Northeast", "Great Lakes", "Plains", "West"
]

for region in regions:
    org25 = data25[
        (data25["Geography"] == region) &
        (data25["Period"] <= 32) &
        (data25["Type"] == "Organic")
    ]["Total Bulk and Bags"].sum()

    org26 = data26[
        (data26["Geography"] == region) &
        (data26["Period"] <= 32) &
        (data26["Type"] == "Organic")
    ]["Total Bulk and Bags"].sum()

    print(region, "organic growth %:", pct_change(org25, org26))
```
